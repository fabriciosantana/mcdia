```python
import unicodedata
```

```python
original = 'Café moído de graça pelo vovô'
print(original)
```

```python
fraseDecomposta = unicodedata.normalize("NFD", original)
print(fraseDecomposta)
```

```python
ascii = fraseDecomposta.encode("ascii", "ignore")
ascii
```

```python
utf8 = ascii.decode("utf-8").upper()
utf8
```
