1. Importar pacotes

```python
import pandas as pd
import requests
```

2. Definir a URL

```python
server_url = "https://dadosabertos.compras.gov.br/modulo-material/2_consultarClasseMaterial"
```

3. Fazer a requisição para obter os dados

```python
pagina = 1
paginas_restantes = 1
resultado_classe = []

while paginas_restantes != 0:
    
    response = requests.get(server_url, params={'pagina': pagina})

    if response.status_code == 200:
        json_response = response.json()
        print(json_response)
        resultado_classe.extend(json_response["resultado"])
        paginas_restantes = json_response["paginasRestantes"]
        pagina = pagina + 1
    else:
        print("Erro: ", response.text)
        paginas_restantes = 0

resultado_classe
```

4. Recuperar o resultado e armazenar em uma variável

```python
df_resultado_classe = pd.DataFrame(resultado_classe)
```

5. Vamos colocar em um Data Frame

```python
df_resultado_classe
```

6. Análise Macro do Data Frame

```python
df_resultado_classe.info()
```

```python
display(df_resultado_classe)

df_resultado_classe['dataHoraAtualizacao'] = pd.to_datetime(df_resultado_classe['dataHoraAtualizacao'])

df_resultado_classe['dataCarga'] = pd.to_datetime('today').normalize()

display(df_resultado_classe)
```
