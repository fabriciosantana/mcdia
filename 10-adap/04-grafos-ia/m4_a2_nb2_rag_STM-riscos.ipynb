```python
from langchain_community.document_loaders import PyPDFLoader
import os

# Define the path to the PDF documents in Google Drive
pdf_folder_path = './grafos/stm'

# List all PDF files in the specified directory
pdf_files = [f for f in os.listdir(pdf_folder_path) if f.endswith('.pdf')]

# Dictionary to store loaded documents, with filename as key
loaded_documents = {}

print(f"Found {len(pdf_files)} PDF files in '{pdf_folder_path}'.")

for pdf_file in pdf_files:
    file_path = os.path.join(pdf_folder_path, pdf_file)
    print(f"Loading document: {pdf_file}...")
    try:
        loader = PyPDFLoader(file_path)
        documents = loader.load()
        # Store the first page of each document or the whole document if preferred
        # For now, we'll store the entire list of pages for each document
        loaded_documents[pdf_file] = documents
        print(f"Successfully loaded {len(documents)} pages from {pdf_file}.")
    except Exception as e:
        print(f"Error loading {pdf_file}: {e}")

print("\nAll specified PDF documents have been processed.")
print(f"Loaded documents keys: {list(loaded_documents.keys())}")
```

Let's inspect the first few pages of one of the loaded documents to verify the content.

```python
if loaded_documents:
    # Get the first document from the dictionary
    first_document_name = list(loaded_documents.keys())[0]
    first_document_pages = loaded_documents[first_document_name]

    print(f"\n--- Content of '{first_document_name}' (first 2 pages) ---")
    for i, page in enumerate(first_document_pages[:2]): # Display first 2 pages
        print(f"Page {i+1}:\n{page.page_content[:500]}...\n") # Print first 500 characters of page content
else:
    print("No documents were loaded.")
```

## Criando uma Base de Conhecimento Vetorial

Para criar uma base de conhecimento vetorial, seguiremos estes passos:
1.  **Instalar bibliotecas necessárias**: `chromadb` e `openai`.
2.  **Dividir documentos**: Dividir os documentos carregados em pedaços menores e gerenciáveis.
3.  **Gerar embeddings**: Converter esses pedaços em representações vetoriais numéricas usando o modelo de embedding da OpenAI.
4.  **Criar o armazenamento vetorial**: Armazenar esses embeddings em uma instância ChromaDB, que permite buscas eficientes por similaridade.

```python
# Install necessary libraries for vector store and embeddings
%pip install -q chromadb openai
```

```python
%pip install -q langchain
%pip install -U langchain-text-splitters
%pip install -q langchain-openai
%pip install -q langchain-community
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
#from google.colab import userdata


# Initialize the text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=10000,
    chunk_overlap=200,
    length_function=len,
    add_start_index=True,
)

# Split documents ensuring chunks do not mix content from different original documents
all_chunks = []
for doc_list in loaded_documents.values():
    # Each doc_list contains pages from a single original PDF
    doc_chunks = text_splitter.split_documents(doc_list)
    all_chunks.extend(doc_chunks)

chunks = all_chunks

print(f"Split {len(loaded_documents)} original documents into {len(chunks)} chunks.")
print("Conteúdo do primeiro chunk:\n", chunks[0].page_content[:500], "...")
print("Metadados do primeiro chunk:\n", chunks[0].metadata)
```

```python
env_path = '.env'

from dotenv import load_dotenv
load_dotenv(env_path)

# Initialize OpenAI Embeddings
embeddings = OpenAIEmbeddings(model="text-embedding-3-large")

# Check if 'db' (the vector database) already exists in memory and will be recreated.
if 'db' in globals() and db is not None:
    print("Base vetorial 'db' existente em memória detectada. Recriando...")

# Create the vector store using ChromaDB. This will overwrite any existing 'db' variable.
db = Chroma.from_documents(chunks, embeddings)

print("Vector database created successfully!")
```

A base de dados vetorial `db` está agora pronta. Você pode usá-la para buscas por similaridade ou para recuperar informações relevantes.

## Testando a Base Vetorial

Agora vamos testar a base vetorial recém-criada, realizando uma busca por similaridade com uma pergunta específica para recuperar o conteúdo relevante.

```python
# Definir a pergunta de consulta
query = "Descreva em linhas gerais o gerenciamento de riscos das contratação na justiça militar da união"

# Realizar a busca por similaridade na base de dados vetorial
# Recuperar os 5 chunks mais relevantes
results = db.similarity_search(query, k=5)

print(f"Recuperados {len(results)} chunks para a pergunta: '{query}'\n")

# Apresentar os chunks recuperados
for i, doc in enumerate(results):
    print(f"--- Chunk {i+1} ---")
    print(f"Conteúdo:\n{doc.page_content[:500]}...") # Exibe os primeiros 500 caracteres
    print(f"Metadados: {doc.metadata}\n")
```

## Realizando Geração Aumentada por Recuperação (RAG)

Agora vamos integrar a base de conhecimento vetorial com um modelo de linguagem para realizar o RAG. Isso permitirá que o modelo de linguagem use o contexto recuperado dos seus documentos para gerar respostas mais precisas e informadas à sua pergunta.

```python
from langchain_openai import ChatOpenAI
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_core.prompts import ChatPromptTemplate
from langchain_classic.chains import create_retrieval_chain


# Initialize the LLM (Large Language Model)
# Using 'gpt-4o' or 'gpt-3.5-turbo' for better performance
llm = ChatOpenAI(model="gpt-5.4", temperature=0.1)

# Create a prompt template for RAG
rag_prompt = ChatPromptTemplate.from_messages([
    ("system", """Você é um assistente útil.
                  Responda à pergunta do usuário com base apenas no contexto fornecido:\n\n{context}
                  Indique os documentos que foram utilizados como base da resposta"""),
    ("user", "{input}"),
])

# Create a chain that combines documents by 'stuffing' them into the prompt
document_chain = create_stuff_documents_chain(llm, rag_prompt)

# Create a retriever from your ChromaDB instance
retriever = db.as_retriever()

# Create the full retrieval chain
retrieval_chain = create_retrieval_chain(retriever, document_chain)

# Invoke the chain with the test query
response = retrieval_chain.invoke({"input": query})

print("--- Resposta do RAG ---")
print(response["answer"])

# You can also inspect the retrieved documents that formed the context
print("\n--- Documentos de Contexto Recuperados ---")
for i, doc in enumerate(response["context"]):
    print(f"Documento {i+1} (Source: {doc.metadata.get('source', 'N/A')} - Page: {doc.metadata.get('page_label', 'N/A')}):\n{doc.page_content[:300]}...\n")
```
