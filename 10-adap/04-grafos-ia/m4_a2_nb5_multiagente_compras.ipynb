### Estado do agente

```python
from typing import TypedDict, List, Optional, Annotated
from langchain_core.messages import BaseMessage, FunctionMessage
from langgraph.graph import add_messages
from pydantic import BaseModel

# Definir o modelo Pydantic para os detalhes do item
class ItemDetalhes(BaseModel):
    CHAVE_COMPRA_PNCP: str
    OBJETO: str
    numero_item: int
    descricao: str
    descricao_detalhada: str
    unidade_fornecimento: str
    valor_estimado: float
    quantidade_solicitada: int
    # Adicionando campos de metadados para melhor rastreamento
    NUMERO_UASG: Optional[str] = None
    NUMERO_COMPRA: Optional[str] = None
    ANO_COMPRA: Optional[int] = None

class ItemNaoSimilar(BaseModel):
    CHAVE_COMPRA_PNCP: str
    OBJETO: str
    numero_item: int
    descricao: str
    descricao_detalhada: str
    unidade_fornecimento: str
    valor_estimado: float
    quantidade_solicitada: int
    # Adicionando campos de metadados para melhor rastreamento
    NUMERO_UASG: Optional[str] = None
    NUMERO_COMPRA: Optional[str] = None
    ANO_COMPRA: Optional[int] = None
    JUSTIFICATIVA_DE_NAO_SIMILARIDADE: Optional[str]

# Definir o estado do agente
class AgentState(TypedDict):
    messages: Annotated[list[BaseMessage], add_messages]
```

### Definição do agente

```python
from langchain.tools import tool
import pandas as pd
from typing import Optional

@tool
def get_pncp_key_by_index(index: int) -> Optional[ItemDetalhes]:
    """Retorna um objeto `ItemDetalhes` de um registro no arquivo 'compras_novas.csv' pelo índice.

    Args:
        index (int): O índice baseado em zero do registro no DataFrame para o qual o objeto `ItemDetalhes` deve ser retornado.

    Returns:
        Optional[ItemDetalhes]: O objeto `ItemDetalhes` correspondente ao índice fornecido, ou `None` se o índice for inválido,
             o arquivo não for encontrado, uma coluna necessária não existir ou ocorrer outro erro.
    """
    #csv_file_path = '/content/drive/MyDrive/Aulas/Aulas IDP/2026/Auditoria de Dados e Accountability/modulo IV - Grafos e IA/compras/compras_novas.csv'
    csv_file_path = 'compras/compras_novas.csv'
    try:
        df_novas_compras = pd.read_csv(
            csv_file_path, 
            sep=";", 
            dtype={
            "CHAVE_COMPRA_PNCP": "string",
            "NUMERO_UASG": "string",
            "NUMERO_COMPRA": "string",}
        )
        if 0 <= index < len(df_novas_compras):
            row = df_novas_compras.loc[index]
            item_detalhes = ItemDetalhes(
                CHAVE_COMPRA_PNCP=row['CHAVE_COMPRA_PNCP'],
                OBJETO=row['OBJETO'],
                numero_item=int(row['numero_item']),
                descricao=row['descricao'],
                descricao_detalhada=row['descricao_detalhada'],
                unidade_fornecimento=row['unidade_fornecimento'],
                valor_estimado=float(row['valor_estimado']),
                quantidade_solicitada=int(row['quantidade_solicitada']),
                NUMERO_UASG= str(row['NUMERO_UASG']),
                NUMERO_COMPRA=str(row['NUMERO_COMPRA']),
                ANO_COMPRA=int(row['ANO_COMPRA'])
            )
            return item_detalhes
        else:
            print(f"Índice {index} fora dos limites do DataFrame.")
            return None
    except FileNotFoundError:
        print(f"Arquivo não encontrado: {csv_file_path}")
        return None
    except KeyError as e:
        print(f"Coluna não encontrada no arquivo CSV ou nome incorreto: {e}")
        return None
    except Exception as e:
        print(f"Ocorreu um erro ao carregar ou processar o arquivo: {e}")
        return None
```

```python
@tool
def calculate_median_price(prices: list[float]) -> float:
    """Calcula a mediana de uma lista de preços.

    Args:
        prices (list[float]): Uma lista de valores numéricos representando os preços.

    Returns:
        float: O valor da mediana dos preços fornecidos. Retorna None se a lista for vazia.
    """
    if not prices:
        return None
    import statistics
    return statistics.median(prices)
```

### Realizando Geração Aumentada por Recuperação (RAG)

Agora vamos integrar a base de conhecimento vetorial com um modelo de linguagem para realizar o RAG. Isso permitirá que o modelo de linguagem use o contexto recuperado dos seus documentos para gerar respostas mais precisas e informadas à sua pergunta.

```python
from langchain.tools import tool
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough, RunnableLambda
from typing import List # Ensure List is imported
from operator import itemgetter

# Initialize the LLM (Large Language Model)
llm = ChatOpenAI(model="gpt-5.4", temperature=0.1)

# Create a retriever from your ChromaDB instance, returning 15 items
retriever = db.as_retriever(search_kwargs={'k': 15})

# 1. Definir o modelo Pydantic para a lista de saída estruturada
class ListaDeItens(BaseModel):
    """Representa uma lista de itens de compra analisados encontrados na base de conhecimento."""
    itens_similares: List[ItemDetalhes]
    itens_nao_similares: List[ItemNaoSimilar]

# 2. Configurar o LLM para retornar saída estruturada
structured_llm = llm.with_structured_output(ListaDeItens)

# Helper function to format retrieved documents into a single string, including metadata
def format_docs_with_metadata(docs):
    formatted_docs = []
    for doc in docs:
        content = doc.page_content
        metadata = doc.metadata
        formatted_docs.append(f"--- Documento de Contexto ---\nConteúdo: {content}\nMetadados: {metadata}\n--------------------------")
    return "\n\n".join(formatted_docs)

# 3. Criar um prompt específico para RAG com saída estruturada
rag_prompt_structured = ChatPromptTemplate.from_messages([
    ("system", """Você é um assistente especialista em compras públicas. Sua tarefa é encontrar itens de compra semanticamente similares ao 'item_alvo' fornecido,
                  EXCLUSIVAMENTE dentro do contexto de compras públicas disponibilizado. Para cada item similar encontrado no contexto, extraia todas as informações necessárias para preencher o modelo ItemDetalhes.
                  Os metadados dos documentos de contexto estão disponíveis na seção 'Metadados:' de cada 'Documento de Contexto'.
                  Certifique-se de extrair `CHAVE_COMPRA_PNCP`, `NUMERO_UASG`, `NUMERO_COMPRA` e `ANO_COMPRA` dos metadados dos documentos de contexto, se disponíveis.
                  Voce deve retornar 2 listas ESTRITAMENTE seguindo o esquema `ListaDeItens`:
                    A: lista de itens similares - dos itens presentes no contexto, relacione aqui aqueles itens que são
                       similares ao item alvo.
                    B: Lista de itens não similares - dos itens presentes no contexto, relacione aqui aqueles que não sao similares para fins de cmparação de preço.    DOS ESTRITAMENTE seguindo o esquema `ListaDeItens`.
                  Não inclua nenhum item que não esteja explicitamente no contexto.
                  Se nenhum item similar for encontrado, retorne uma lista vazia.
                  Contexto de compras (com conteúdo e metadados):\n{context}
               """),
    ("user", "Com base no contexto, encontre todos os itens de compra similares ao seguinte item alvo: {input}"),
])

# 4. Criar uma cadeia de documentos estruturada usando LCEL
chain_structured = (
     rag_prompt_structured # Pass formatted context and input to the prompt
    | structured_llm # Pass the prompted input to the structured LLM
)

# A `retrieval_chain_structured` será removida e a lógica de recuperação será feita explicitamente na ferramenta.

@tool
def find_similar_items(item_alvo: dict) -> ListaDeItens:
    """Encontra itens de compra similares a um item alvo especificado, utilizando uma base de conhecimento vetorial.

    Args:
        item_alvo (dict): O item alvo para o qual buscar itens similares, contendo as informações necessárias:
                                  CHAVE_COMPRA_PNCP, OBJETO, descricao, descricao_detalhada, unidade_fornecimento, valor_estimado e quantidade_solicitada.

    Returns:
        ListaDeItens: Um objeto contendo uma lista de objetos ItemDetalhes que são considerados similares ao item alvo ou não,
                               extraídos da base de conhecimento. Retorna uma lista vazia se nenhum item similar for encontrado
                               ou em caso de erro.
    """
    # Construir a string de consulta para a cadeia RAG com base nos atributos do item_alvo
    query_string = \
        f"Item alvo para busca de similares: Objeto: {item_alvo["OBJETO"]}, " +\
        f"Descrição: {item_alvo["descricao"]}, "+\
        f"Descrição Detalhada: {item_alvo["descricao_detalhada"]}, "+\
        f"Unidade de Fornecimento: {item_alvo["unidade_fornecimento"]}. " +\
        f"Valor Estimado: {item_alvo["valor_estimado"]}, Quantidade Solicitada: {item_alvo["quantidade_solicitada"]}."


    try:
        #print("Query para o retriever: <"+query_string+">")
        # 1. Recuperar documentos explicitamente
        retrieved_docs = retriever.invoke(query_string)
        #print(f"Documentos recuperados: {len(retrieved_docs)}")

        # 2. Formatar os documentos com metadados
        formatted_docs = format_docs_with_metadata(retrieved_docs)
        #print("Documentos formatados:")
        #print(formatted_docs)

        # 3. Invocar a cadeia de documentos estruturada com os documentos recuperados e a query
        response = chain_structured.invoke({"context": formatted_docs, "input": query_string})
        #print("Response do structured_llm: <"+str(response)+">")

        # A 'response' já é o objeto ListaDeItensSimilares diretamente do structured_llm
        return response
    except Exception as e:
        print(f"Erro ao buscar itens similares: {e}")
        return ListaDeItens(itens_similares=[])
```

## Construindo o Agente de Análise de Preços

```python
from typing import TypedDict, Annotated, List, Optional
from langchain_core.messages import BaseMessage, FunctionMessage, HumanMessage, AIMessage
from langchain_core.runnables import RunnablePassthrough
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode # Remover ToolExecutor da importação

# As classes ItemDetalhes, AgentState e as ferramentas (get_pncp_key_by_index, calculate_median_price, find_similar_items)
# são definidas em células anteriores e devem estar no escopo.
# O 'llm' e o 'retriever' também devem estar definidos em células anteriores.

# 1. Agrupar as ferramentas
tools_check_overpricing = [
    get_pncp_key_by_index,
    calculate_median_price,
    find_similar_items
]

# 2. Criar um LLM com capacidade de chamada de ferramentas
agent_llm = llm.bind_tools(tools_check_overpricing)

# 3. Definir o nó do agente que invoca o LLM
def agent_check_overpricing(state: AgentState):
    system_prompt = """Você é um agente especializado em auditoria de dados de compras públicas. Seu objetivo é ajudar o usuário a analisar o valor de um item alvo em comparação com preços medianos de itens similares.
                        Siga estas etapas:
                        1.  **Obter o Item Alvo**: O usuário fornecerá um índice para o item alvo. Use a ferramenta `get_pncp_key_by_index` para recuperar os detalhes completos desse item.
                        2.  **Encontrar Itens de Referência**: Com base nos detalhes do item alvo, use a ferramenta `find_similar_items` para encontrar uma lista de itens de compra similares na base de conhecimento.
                        3.  **Calcular Mediana**: Extraia os valores estimados (`valor_estimado`) dos `itens_referencia` e use a ferramenta `calculate_median_price` para calcular a mediana dos preços.
                        4.  **Comparar e Reportar**: Compare o `valor_estimado` do `item_alvo` com a mediana calculada e determine se o preço do item alvo está acima, abaixo ou igual à mediana.
                        5.  **Comunicar o Resultado**: Apresente um resumo claro e conciso ao usuário, informando o item alvo, incluindo a CHAVE_COMPRA_PNCP, a mediana dos preços dos itens similares, a lista de produtos similares incluindo a UASG e o ANO da compra, o mumero
                            da compra e o mumero do item e a comparação (acima/abaixo/igual).
                        6.  **Lista de Itens não similares**: Inclua em sua resposta a lista de itens que foram
                            fornecidos no contexto, mas não foram considerados similares com a justificativa
                        Seja objetivo e direto. Se alguma ferramenta falhar ou não encontrar dados, informe o usuário e tente prosseguir com o que for possível ou finalize se não houver dados para continuar a análise.
                       """
    prompt_template = ChatPromptTemplate.from_messages([
        ("system", system_prompt),
        ("placeholder", "{messages}"),
    ])

    chain = prompt_template | agent_llm

    messages = state['messages']
    # O agente invoca o LLM com as mensagens atuais. ToolNode cuidará dos resultados das ferramentas.
    response = chain.invoke({"messages": messages})
    #print(response)
    return {"messages": [response]}

```

```python
# Parser auxiliar
def pncp_id_parser(pncp_id: str) -> tuple[str, str, str]:
    """Parses a PNCP ID string to extract CNPJ, sequential number, and year.

    The PNCP ID is expected to be in the format: CNPJ (13 digits) + fixed '1' (1 digit) + sequential (6 digits) + year (4 digits).

    Args:
        pncp_id (str): The PNCP ID string to parse. It should be a 24-character string.

    Returns:
        tuple[str, str, str]: A tuple containing three strings: (cnpj, sequential, year).
                              Returns (None, None, None) if the format does not match the expected pattern.
    """
    # Expected format: CNPJ (14) + fixed '1' (1) + sequential (6) + year (4)
    # Total length = 14 + 1 + 6 + 4 = 25 characters
    if not isinstance(pncp_id, str) or len(pncp_id) != 25 or pncp_id[14] != '1':
        print(f"Invalid PNCP ID format: {pncp_id}. Expected 24 characters with '1' at position 13.")
        return None, None, None

    cnpj = pncp_id[0:14]
    sequential = pncp_id[15:21]
    year = pncp_id[21:25]

    return cnpj, sequential, year
```

```python
# funcao auxiliar
def get_file_links(pncp_id:str) -> list[str]:
    cnpj_orgao, sequencial_compra, ano_compra = pncp_id_parser(pncp_id)
    api_url = f"https://pncp.gov.br/api/pncp/v1/orgaos/{cnpj_orgao}/compras/{ano_compra}/{sequencial_compra}/arquivos"

    try:
        # Os parâmetros devem estar incluídos diretamente na files_api_url, não como um dicionário 'params' separado.
        response_files = requests.get(api_url)
        response_files.raise_for_status()
        files_data = response_files.json()

    except requests.exceptions.RequestException as e:
        print(f"Erro ao buscar links de arquivos: {e}")
        raise e

    return files_data
```

```python
import requests
import os
import re
import urllib.parse
import zipfile # Import the zipfile module

# funcao auxiliar
def download_files(pncpid, files: list):
    """Baixa arquivos de uma lista de URLs para a pasta 'arquivos'."""
    download_folder = os.path.join("arquivos", pncpid)
    os.makedirs(download_folder, exist_ok=True)

    print(f"Iniciando download de {len(files)} arquivos...")
    for i, file_data in enumerate(files):
        try:
            response = requests.get(file_data["url"], stream=True)
            response.raise_for_status() # Raise HTTPError for bad responses (4xx or 5xx)

            file_path = os.path.join(download_folder, file_data["titulo"].replace("/","_")) # Use 'titulo' as suggested by user interaction
            file_data.update({"file_path": file_path})
            with open(file_path, 'wb') as f:
                for chunk in response.iter_content(chunk_size=8192):
                    f.write(chunk)
            print(f"[{i+1}/{len(files)}] Download concluído: ({file_data["tipoDocumentoId"]})-{file_data["tipoDocumentoNome"]}: {file_data["titulo"]} ")

            # Check if the downloaded file is a zip file and unzip it
            if file_path.lower().endswith('.zip'):
                print(f"[{i+1}/{len(files)}] Detectado arquivo ZIP. Descompactando {os.path.basename(file_path)}...")
                try:
                    with zipfile.ZipFile(file_path, 'r') as zip_ref:
                        zip_ref.extractall(download_folder)
                    print(f"[{i+1}/{len(files)}] Arquivo ZIP descompactado com sucesso em: {download_folder}")
                    os.remove(file_path) # Remove the zip file after extraction
                    print(f"[{i+1}/{len(files)}] Arquivo ZIP original removido: {file_path}")
                except zipfile.BadZipFile:
                    print(f"[{i+1}/{len(files)}] Erro: O arquivo {os.path.basename(file_path)} não é um arquivo ZIP válido.")
                except Exception as unzip_e:
                    print(f"[{i+1}/{len(files)}] Erro ao descompactar {os.path.basename(file_path)}: {unzip_e}")

        except requests.exceptions.RequestException as e:
            print(f"[{i+1}/{len(files)}] Erro ao baixar {file_data["url"]}: {e}")
        except Exception as e:
            print(f"[{i+1}/{len(files)}] Ocorreu um erro inesperado ao processar {file_data["url"]}: {e}")
    print("Todos os downloads foram processados.")
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
#from langchain_community.vectorstores import Chroma
from langchain_chroma import Chroma
from langchain_core.documents import Document
from typing import List
import os
import glob
from dotenv import load_dotenv

# Ensure environment variables are loaded for OpenAI API key
#env_path = '/content/drive/MyDrive/Aulas/Aulas IDP/2026/Auditoria de Dados e Accountability/modulo IV - Grafos e IA/.env'
env_path = '.env'
load_dotenv(env_path)

@tool
def get_chunks(pncp_id: str, query: str) -> List[Document]:
    """Loads PDF documents of a given pncp_id, splits them into chunks,
    creates a ChromaDB vector store, and performs a similarity search based on a query.

    Args:
        pncp_id (str): The pncp_id for which is to be retrieved the chunks of PDF files.
        query (str): The query string to search for similar chunks.

    Returns:
        List[Document]: A list of Document objects that are semantically similar to the query.
                        Returns an empty list if no documents are loaded or no similar chunks are found.
    """
    documents = []
    file_path = os.path.join("arquivos", pncp_id)
    # Determine if file_path is a directory or a single file
    if os.path.isdir(file_path):
        # Load all PDF files from the directory (recursively)
        pdf_files = glob.glob(os.path.join(file_path, '**/*.pdf'), recursive=True)
        if not pdf_files:
            print(f"No PDF files found in directory: {file_path}")
            return []
        for pdf_file in pdf_files:
            try:
                loader = PyPDFLoader(pdf_file)
                documents.extend(loader.load())
            except Exception as e:
                print(f"Error loading PDF {pdf_file}: {e}")
    elif os.path.isfile(file_path) and file_path.lower().endswith('.pdf'):
        # Load a single PDF file
        try:
            loader = PyPDFLoader(file_path)
            documents.extend(loader.load())
        except Exception as e:
            print(f"Error loading PDF {file_path}: {e}")
    else:
        print(f"Invalid file_path: {file_path}. Must be a PDF file or a directory containing PDFs.")
        return []

    if not documents:
        print("No documents were loaded for chunking and search.")
        return []

    # Split documents into chunks
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=250,
        chunk_overlap=50,
        length_function=len
    )

    for doc in documents:
        doc.page_content = doc.page_content.upper()

    chunks = text_splitter.split_documents(documents)
    print(f"Divided {len(documents)} documents into {len(chunks)} chunks.")

    # Initialize OpenAI Embeddings
    embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

    # Create a vector store from the chunks (in-memory for this tool call)
    vectorstore = Chroma.from_documents(chunks, embeddings,persist_directory="vectordb_2")
    print("Vector store created from chunks.")

    #print(f"Query:{query}")

    # Perform similarity search
    retrieved_chunks = vectorstore.similarity_search(query.upper(), k=10) # Retrieve top 10 similar chunks
    print(f"Found {len(retrieved_chunks)} similar chunks for the query.")

    #for i, c in enumerate(retrieved_chunks):
    #    print(f"\n ------------ \nDocumento {i}: {c.page_content}\n")

    return retrieved_chunks
```

## Construção do agente verificador das descrições dos itens

```python
# 1. Agrupar as ferramentas
tools_check_item_specifications = [
    get_pncp_files,
    get_chunks
]

# 2. Criar um LLM com capacidade de chamada de ferramentas
agent_check_item_specifications_llm = llm.bind_tools(tools_check_item_specifications)

# 3. Definir o nó do agente que invoca o LLM
def agent_check_item_specifications(state: AgentState):
    system_prompt = """Você é um agente especializado em auditoria de dados de compras públicas.
                       Seu objetivo é dar continuidade a uma audititoria iniciada por um agente anterior que já realizou
                       uma análise de preços e agora verificar se a especificação do item alvo encontrada nos metadados do PNCP
                       é equivalente à especificação constante nos documentos da contratação.
                       Para isso você deve:
                       1. **Obter o Item Alvo**: O item alvo já está presente no histórico de mensagens realizado pelo agente anterior.
                          Um dado importante do item alvo é a CHAVE_COMPRA_PNCP, que é equivalente ao id_pncp que é usado nas ferramentas.
                          Para obter o item alvo, você deve examinar as mensagens anteriores do `HumanMessage` que iniciou a análise.
                       2. **Obter os arquivos da compra**: com base no item alvo recuperado do histórico de mensagens, executar a
                          ferramenta `get_pncp_files` passando o pncp_id que corresponde à CHAVE_COMPRA_PNCP.
                       3. **Obter os trechos dos documentos da contratação**: usar a ferramenta `get_chunks` para obter trechos
                          relevantes dos documentos da contratação. A ferramenta get_chunks espera receber
                          um pncp_id e uma query que deve corresponder à descricao do item alvo.
                       4. de posse dos chunks, você deve verificar se a descrição do item alvo presente
                          nos metadados é equivalente à mais completa das descrições presentes nos trechos dos
                          documentos. Caso os trechos dos documentos apresentem especificações mais
                          completas, você deve apontar que existe um problema nos metadados do PNCP, pois a
                          descrição do metadados é insuficiente para possibilitar uma boa avaliação de preços, já
                          que os fornecedores se baseam nos documentos para fazer suas cotações.

                          ATENÇÃO: OS passos devem ser executados em sequencia. Não executar em paralelo.
                          É necessário concluir a execuçao da ferramenta `get_pncp_files` antes de
                          chamar a ferramenta `get_chunks`
                          """

    prompt_template = ChatPromptTemplate.from_messages([
        ("system", system_prompt),
        ("placeholder", "{messages}"),
    ])

    chain = prompt_template | agent_check_item_specifications_llm

    messages = state['messages']
    # The agent invokes the LLM with the current messages. ToolNode will handle tool results.
    response = chain.invoke({"messages": messages})
    return {"messages": [response]}
```

## Monta o workflow multiagente

```python
# 5. Definir a lógica condicional para decidir o próximo passo
def should_continue_check_overpricing(state: AgentState) -> str:
    messages = state['messages']
    last_message = messages[-1]
    # Se a última mensagem do agente inclui chamadas de ferramentas, vá para 'call_tool'.
    # Caso contrário, o agente chegou a uma conclusão e deve terminar.
    if last_message.tool_calls:
        return "call_tool_check_overpricing"
    else:
        return "agent_check_item_specifications"

# 6. Construir o StateGraph
workflow = StateGraph(AgentState)

workflow.add_node("agent_check_overpricing_node", agent_check_overpricing)
workflow.add_node("agent_check_item_specifications_node", agent_check_item_specifications)

# Usar ToolNode padrão para executar as ferramentas
workflow.add_node("call_tool_check_overpricing", ToolNode(tools=tools_check_overpricing))

workflow.set_entry_point("agent_check_overpricing_node")

workflow.add_conditional_edges(
    "agent_check_overpricing_node",
    should_continue_check_overpricing,
    {
        "call_tool_check_overpricing": "call_tool_check_overpricing",
        "agent_check_item_specifications": "agent_check_item_specifications_node"
    },
)
# Após a execução da ferramenta, o fluxo retorna ao agent_node para o LLM processar o resultado da ferramenta
workflow.add_edge('call_tool_check_overpricing', 'agent_check_overpricing_node')

workflow.add_node("call_tool_check_item_specifications", ToolNode(tools=tools_check_item_specifications))

def should_continue_check_item_specifications(state: AgentState) -> str:
    messages = state['messages']
    last_message = messages[-1]
    # Se a última mensagem do agente inclui chamadas de ferramentas, vá para 'call_tool'.
    # Caso contrário, o agente chegou a uma conclusão e deve terminar.
    if last_message.tool_calls:
        return "call_tool_check_item_specifications"
    else:
        return "end_agent"

workflow.add_conditional_edges(
    "agent_check_item_specifications_node",
    should_continue_check_item_specifications, # Corrected: use should_continue_check_item_specifications
    {
        "call_tool_check_item_specifications": "call_tool_check_item_specifications",
        "end_agent": END
    },
)
# Após a execução da ferramenta, o fluxo retorna ao agent_node para o LLM processar o resultado da ferramenta
workflow.add_edge('call_tool_check_item_specifications', 'agent_check_item_specifications_node')

# 7. Compilar o gráfico em um agente executável
agent_executor = workflow.compile()
```

### Executando o Agente

Agora você pode interagir com o agente fornecendo um índice para o item alvo. O agente usará suas ferramentas para realizar a análise e retornar o resultado.

```python
def run_agent_analysis(item_index: int):
    # Initialize the state for the agent with the user's request
    # The system message is already part of the agent_prompt defined above.
    # The agent_runner expects a list of messages for its input.
    # So, we start with a HumanMessage requesting the item at the given index.

    initial_input = {
        "messages": [HumanMessage(content=f"Analise o item com índice {item_index}.")]
    }

    # The AgentState needs to be correctly initialized for the graph.
    # The initial_input will be merged into the AgentState at the start.

    print(f"Iniciando análise para o item com índice: {item_index}")

    # Stream events from the agent to see its steps
    for s in agent_executor.stream(initial_input):
        if "agent_check_item_specifications_node" in s:
            for message in s["agent_check_item_specifications_node"]["messages"]:
                print(f"Agent2: {message.content}")
                try:
                    if message.tool_calls:
                        for tc in message.tool_calls:
                            # Acessar 'name' e 'args' como chaves de dicionário, pois 'tc' parece ser um dicionário
                            print(f"Agent calls tool: {tc['name']} with args {tc['args']}")
                except:
                  pass
        if "agent_check_overpricing_node" in s:
            for message in s["agent_check_overpricing_node"]["messages"]:
                print(f"Agent1: {message.content}")
                if message.tool_calls:
                    for tc in message.tool_calls:
                        # Acessar 'name' e 'args' como chaves de dicionário, pois 'tc' parece ser um dicionário
                        print(f"Agent calls tool: {tc['name']} with args {tc['args']}")
        elif "call_tool" in s:
            for message in s["call_tool"]["messages"]:
                print(f"Tool Output: {message.content}")
        elif END in s:
            final_state = s[END]
            final_message = final_state['messages'][-1]
            print("\n--- Análise Finalizada ---")
            print(final_message.content)


# Exemplo de uso do agente:
# run_agent_analysis(item_index=0) # Altere o índice conforme necessário para testar
```
