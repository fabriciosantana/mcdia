### Introdução aos Prompt Templates

### Introdução às Chains (Cadeias)

Uma cadeia permite que você defina uma sequência de passos onde a saída de um passo se torna a entrada para o próximo. Isso é fundamental para construir aplicações de IA generativa robustas e personalizáveis.

### Explicação do `StrOutputParser`

O `StrOutputParser` é um componente do LangChain que simplifica o processamento da saída de um Large Language Model (LLM). Sua principal função é converter a saída gerada pelo LLM em uma string pura. Muitos LLMs retornam objetos complexos ou formatos de mensagem que podem não ser diretamente o texto final que você deseja usar. O `StrOutputParser` extrai o conteúdo de texto principal desse objeto de saída, tornando-o fácil de usar em outras partes da sua aplicação ou para exibição.

No contexto das chains, ele atua como o passo final, garantindo que o resultado da cadeia seja uma string formatada e legível.

### Usando `JsonOutputParser` para Saídas Estruturadas

Em muitas aplicações, não queremos apenas uma string de texto como saída do LLM, mas sim dados estruturados, como JSON. O `JsonOutputParser` do LangChain é ideal para isso, pois ele tenta analisar a saída do LLM como JSON, permitindo que você trabalhe com os dados de forma programática.

Vamos criar um exemplo onde o LLM responderá a uma pergunta sobre a Copa do Mundo em um formato JSON específico, contendo o time vencedor, o país sede e o estádio da final.

```python
from langchain_core.output_parsers import JsonOutputParser
from pydantic import BaseModel, Field
from langchain_core.prompts import PromptTemplate

# 1. Defina o esquema da saída JSON esperada
class WorldCupInfo(BaseModel):
    vencedor: str = Field(description="O time que venceu a Copa do Mundo no ano especificado.")
    sede: str = Field(description="O país onde a Copa do Mundo foi sediada.")
    estadio_final: str = Field(description="O nome do estádio onde a partida final foi jogada.")
    ano: int = Field(description="O ano da Copa do Mundo.")

# 2. Crie uma instância do parser JSON
parser = JsonOutputParser(pydantic_object=WorldCupInfo)

# 3. Crie o Prompt Template com as instruções para o LLM gerar JSON
prompt = PromptTemplate(
    template="""O Usuário vai fornecer um ano de ocorrencia de uma copa do mundo.
                Sua resposta DEVE ser em formato JSON de acordo com o
                seguinte esquema:\n{format_instructions}\n\nPergunta: {query}\n""",
    input_variables=["query"],
    partial_variables={
        "format_instructions": parser.get_format_instructions()
    },
)

# 4. Crie a cadeia: PromptTemplate | LLM | JsonOutputParser
json_chain = prompt | llm | parser

# 5. Execute a cadeia com o ano
query = "1998"
response_json = json_chain.invoke({"query": query})

# 6. Imprima a resposta JSON
print(response_json)
```

Este exemplo demonstra como você pode usar o `JsonOutputParser` para garantir que a saída do seu LLM esteja em um formato JSON bem definido, o que é crucial para integrar LLMs em sistemas que dependem de dados estruturados.

### Entendendo Pydantic e BaseModel

No exemplo acima, utilizamos `Pydantic` e seu `BaseModel` para definir o esquema do JSON esperado. O `BaseModel` é uma classe base que permite criar modelos de dados com validação e parsing de dados automáticos. Cada atributo da classe (`vencedor`, `sede`, `estadio_final`) é tipado (e.g., `str`) e pode ter um `Field` associado, que fornece metadados adicionais, como uma `description`. Esta descrição é especialmente útil, pois o `JsonOutputParser` a utiliza para gerar as instruções de formatação que são passadas para o LLM, ajudando-o a entender como deve ser a estrutura da saída JSON.

### Visualizando as `format_instructions`

Antes de ajustar o prompt, vamos ver o que o método `parser.get_format_instructions()` retorna. Essas são as instruções que o LLM recebe para entender o formato JSON esperado, baseado no seu `Pydantic BaseModel`:

```python
print(parser.get_format_instructions())
```

### Ajustando o Prompt com um Exemplo de JSON Explícito

Para tornar a especificação do formato ainda mais robusta e explícita para o LLM, podemos incluir um exemplo de JSON diretamente no `PromptTemplate`. Isso serve como um modelo adicional para o LLM, orientando-o sobre a estrutura exata que deve ser gerada.

Note que estamos adicionando um `JSON_EXAMPLE` ao nosso template.

```python
from langchain_core.prompts import PromptTemplate

# Defina um exemplo de JSON para incluir no prompt
JSON_EXAMPLE = """
{
  "vencedor": "Brasil",
  "sede": "Estados Unidos",
  "estadio_final": "Rose Bowl"
}
"""

# Crie o Prompt Template com as instruções e o exemplo JSON
# A instrução 'Your response must be a JSON object, like the example below:' é nova.
prompt_with_example = PromptTemplate(
    template="""O Usuário vai fornecer um ano de ocorrencia de uma copa do mundo.
                Sua resposta DEVE ser um objeto JSON, como o exemplo abaixo:
                {json_example}
                Sua resposta DEVE seguir o seguinte esquema:
                {format_instructions}

                Pergunta: {query}
""",
    input_variables=["query"],
    partial_variables={
        "format_instructions": parser.get_format_instructions(),
        "json_example": JSON_EXAMPLE
    },
)

# Crie a cadeia com o novo prompt
json_chain_with_example = prompt_with_example | llm | parser

# Execute a cadeia com o ano
query = "2014"
response_json_with_example = json_chain_with_example.invoke({"query": query})

# Imprima a resposta JSON
print(response_json_with_example)
```

Ao incluir um `JSON_EXAMPLE` diretamente no prompt, estamos dando ao LLM um guia mais visual e explícito sobre o formato desejado, o que pode aumentar a confiabilidade da saída JSON, especialmente para modelos menores ou tarefas mais complexas.

## Introdução a Embeddings e Bancos Vetoriais

### O que são Embeddings?

**Embeddings** são representações numéricas (vetores) de texto, imagens, áudio ou outros tipos de dados, em um espaço de alta dimensão. A ideia principal é que itens com significados semelhantes ou características parecidas estarão mais próximos uns dos outros neste espaço vetorial. Por exemplo, a palavra 'cachorro' e 'cão' teriam vetores muito próximos, enquanto 'mesa' estaria mais distante.

Para texto, os LLMs e modelos de linguagem são excelentes em gerar esses embeddings, capturando o contexto e o significado semântico das palavras e frases.

### O que são Bancos Vetoriais?

Um **Banco Vetorial** (ou *Vector Database*) é um tipo de banco de dados otimizado para armazenar, indexar e pesquisar vetores de alta dimensão de forma eficiente. Diferente dos bancos de dados tradicionais que buscam por correspondência exata de valores, os bancos vetoriais permitem a **busca de similaridade**, encontrando vetores que são 'próximos' ao vetor de consulta no espaço vetorial.

Isso é fundamental para aplicações de IA como:
*   **Busca Semântica:** Encontrar documentos ou informações com base no significado, e não apenas nas palavras-chave.
*   **Recomendação:** Sugerir itens semelhantes com base no histórico ou preferências do usuário.
*   **Geração Aumentada por Recuperação (RAG):** Fornecer informações relevantes para um LLM gerar respostas mais precisas e contextuais.

### Como funcionam?

1.  **Criação de Embeddings:** O texto (ou outro dado) é transformado em um vetor numérico usando um modelo de embedding.
2.  **Armazenamento no Banco Vetorial:** Esses vetores são armazenados no banco, que os indexa para buscas eficientes.
3.  **Consulta:** Quando uma consulta é feita, ela também é convertida em um vetor, e o banco vetorial encontra os vetores armazenados mais próximos (mais semelhantes) usando métricas de distância (como a distância cosseno).

```python
import os

# Instale as bibliotecas necessárias para embeddings e um banco vetorial simples (FAISS)
%pip install --quiet sentence-transformers faiss-cpu langchain-community

from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings
from langchain_community.embeddings import SentenceTransformerEmbeddings

# 1. Defina alguns documentos de exemplo
documents = [
    "O Brasil é o maior país da América do Sul e o quinto maior do mundo.",
    "A capital do Brasil é Brasília.",
    "A culinária brasileira é rica e diversa, com pratos como feijoada e moqueca.",
    "A França é famosa por sua culinária, vinhos e a Torre Eiffel.",
    "Paris é a capital da França e um importante centro cultural e econômico.",
    "A Copa do Mundo de 2014 foi realizada no Brasil e a Alemanha foi a campeã."
]

# 2. Inicialize o modelo de embeddings
# Usaremos OpenAIEmbeddings, mas você pode usar SentenceTransformerEmbeddings para modelos locais

# Verifica se a variável de ambiente OPENAI_API_KEY está definida
if os.getenv("OPENAI_API_KEY") is not None:
    embeddings_model = OpenAIEmbeddings()
    print("Usando OpenAIEmbeddings.")
else:
    print("OPENAI_API_KEY não encontrada. Usando SentenceTransformer para embeddings locais. Isso pode demorar para baixar o modelo.")
    # Se não tiver a chave da OpenAI, use um modelo local da Sentence Transformers
    embeddings_model = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")

# 3. Crie o banco de dados vetorial FAISS a partir dos documentos e embeddings
# A função 'from_texts' cria os embeddings e o índice FAISS automaticamente
vector_db = FAISS.from_texts(documents, embeddings_model)

print("Banco de dados vetorial FAISS criado com sucesso!")
```

```python
# 4. Realize uma busca de similaridade (busca semântica)

query = "Qual é a capital do país conhecido por sua torre famosa?"

# Use 'similarity_search' para encontrar os documentos mais relevantes para a consulta
results = vector_db.similarity_search(query, k=3) # k=2 retorna os 2 documentos mais similares

print(f"Consulta: {query}\n")
print("Resultados da busca semântica:")
for i, doc in enumerate(results):
    print(f"Documento {i+1}: {doc.page_content}")
```

Neste exemplo, você pode ver que a busca semântica retornou documentos relacionados ao significado da sua pergunta, mesmo que as palavras exatas ('torre famosa') não estivessem presentes em um único documento, mas o conceito de 'Torre Eiffel' e 'França' foi capturado pelos embeddings e encontrado no banco vetorial.

## Introdução a Document Readers

**Document Loaders** são componentes do LangChain que permitem carregar dados de diversas fontes (arquivos PDF, HTML, JSON, bancos de dados, etc.) e convertê-los em um formato padronizado de 'Documentos' que o LangChain pode processar. Cada 'Documento' geralmente contém o `page_content` (o texto extraído) e `metadata` (informações adicionais como número da página, fonte, etc.).

Vamos usar o `PyPDFLoader` para carregar um arquivo PDF.

```python
# Instale a biblioteca necessária para leitura de PDFs
%pip install --quiet pypdf

from langchain_community.document_loaders import PyPDFLoader

# Caminho para o arquivo PDF de exemplo definido anteriormente
pdf_exemplo = '/content/drive/MyDrive/Aulas/Aulas IDP/2026/IA Generativa/Aula 1/receitas/Bolo de mandioca com coco na Air Fryer da Rita Lobo.pdf'

# Inicialize o loader com o caminho do arquivo
loader = PyPDFLoader(pdf_exemplo)

# Carregue o documento PDF
# Isso dividirá o PDF em documentos, geralmente um por página
docs = loader.load()

print(f"Total de páginas carregadas: {len(docs)}")

# Exiba o conteúdo da primeira página (ou de algumas páginas para ter uma ideia)
if docs:
    print("\nConteúdo da primeira página:")
    print(docs[0].page_content[:500]) # Exibe os primeiros 500 caracteres
    print("\nMetadata da primeira página:")
    print(docs[0].metadata)
else:
    print("Nenhum documento carregado.")
```

Como você pode ver, o `PyPDFLoader` extraiu o texto de cada página do PDF e o armazenou como um `Document` separado, incluindo metadados úteis como o número da página e a fonte do arquivo.

## Carregando Múltiplos Documentos PDF e Criando um Banco Vetorial

Agora, vamos aplicar o conceito de Document Readers para carregar *todos* os arquivos PDF de uma pasta específica. Em seguida, usaremos esses documentos para popular um banco vetorial, o que nos permitirá fazer buscas semânticas em todo o conteúdo da pasta.

```python
import os
from langchain_community.document_loaders import PyPDFLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import SentenceTransformerEmbeddings
from langchain_openai import OpenAIEmbeddings

# Defina o caminho para a pasta que contém os PDFs
receitas_folder = '/content/drive/MyDrive/Aulas/Aulas IDP/2026/IA Generativa/Aula 1/receitas'

all_docs = []

# 1. Iterar sobre todos os arquivos na pasta
for filename in os.listdir(receitas_folder):
    if filename.endswith('.pdf'):
        file_path = os.path.join(receitas_folder, filename)
        print(f"Carregando: {filename}")
        # Inicialize o loader para cada PDF
        loader = PyPDFLoader(file_path)
        # Carregue os documentos e adicione à lista geral
        all_docs.extend(loader.load())

print(f"\nTotal de páginas carregadas de todos os PDFs: {len(all_docs)}")

# 2. Inicialize o modelo de embeddings (reutilizando o anterior ou inicializando se necessário)
# Verifica se a variável de ambiente OPENAI_API_KEY está definida
if os.getenv("OPENAI_API_KEY") is not None:
    embeddings_model_for_folder = OpenAIEmbeddings()
    print("Usando OpenAIEmbeddings para a pasta de receitas.")
else:
    print("OPENAI_API_KEY não encontrada. Usando SentenceTransformer para embeddings locais na pasta de receitas. Isso pode demorar para baixar o modelo.")
    embeddings_model_for_folder = SentenceTransformerEmbeddings(model_name="all-MiniLM-L6-v2")

# 3. Crie o banco de dados vetorial FAISS a partir de todos os documentos carregados
vector_db_folder = FAISS.from_documents(all_docs, embeddings_model_for_folder)

print("\nBanco de dados vetorial FAISS criado com sucesso para a pasta de receitas!")
```

```python
# 4. Realize uma busca de similaridade no novo banco vetorial da pasta

query_folder = "Quais receitas tem coco?"

# Use 'similarity_search' para encontrar os documentos mais relevantes para a nova consulta
results_folder = vector_db_folder.similarity_search(query_folder, k=3) # Retorna os 3 documentos mais similares

print(f"Consulta: {query_folder}\n")
print("Resultados da busca semântica na pasta de receitas:")
for i, doc in enumerate(results_folder):
    print(f"Documento {i+1} (Fonte: {doc.metadata.get('source', 'Desconhecida')}, Página: {doc.metadata.get('page', 'Desconhecida')}):\n{doc.page_content[:200]}...\n")
```

Este exemplo demonstra a capacidade de carregar um grande número de documentos e indexá-los em um banco vetorial, permitindo buscas semânticas eficientes em todo o seu corpus de dados. Isso é um passo fundamental para construir aplicações de Retrieval Augmented Generation (RAG) robustas.

## Construindo uma Chain de Recuperação e Geração (RAG)

A **Chain de Recuperação e Geração (RAG)** é uma técnica poderosa que combina a capacidade de um Large Language Model (LLM) de gerar texto com a habilidade de recuperar informações relevantes de uma base de dados externa (neste caso, nosso banco vetorial de receitas). Isso permite que o LLM responda a perguntas com base em fatos e informações específicas que não estavam em seu conjunto de treinamento inicial, tornando as respostas mais precisas e atualizadas.

### Como a RAG Chain funciona:

1.  **Recuperação (Retrieval):** Dada uma consulta do usuário, a chain primeiro busca documentos relevantes no nosso `vector_db_folder` usando busca de similaridade.
2.  **Aumento (Augmentation):** Os documentos recuperados são então inseridos no prompt do LLM como 'contexto'.
3.  **Geração (Generation):** O LLM utiliza esse contexto (além de suas próprias capacidades) para gerar uma resposta informada à consulta do usuário.

```python
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from operator import itemgetter # Import itemgetter

# Helper function to format documents into a single string
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# 1. Crie o Retriever a partir do banco vetorial
# O 'as_retriever()' permite buscar documentos relevantes do FAISS
retriever = vector_db_folder.as_retriever()

# 2. Defina o Prompt Template para a Geração
# O prompt incluirá um placeholder para o 'contexto' recuperado
template_rag = """Responda à pergunta com base APENAS no seguinte contexto:
{context}

Pergunta: {question}
"""

rag_prompt = ChatPromptTemplate.from_template(template_rag)

# 3. Crie a Chain RAG
# - Use itemgetter("question") to pass only the question string to the retriever.
# - Chain the retriever's output to format_docs to get a single string context.
# - Pass both the formatted context and the original question to the rag_prompt.
rag_chain = (
    {
        "context": itemgetter("question") | retriever | format_docs,
        "question": itemgetter("question"),
    }
    | rag_prompt
    | llm
    | StrOutputParser()
)

# 4. Defina a pergunta do usuário
user_question = "liste os nomes das receitas que são doces, com o tempo total de preparo!"

# 5. Execute a Chain RAG
response_rag = rag_chain.invoke({"question": user_question})

# 6. Imprima a resposta
print(response_rag)
```

Neste exemplo, a chain RAG primeiro recuperou os documentos relevantes do banco vetorial (`vector_db_folder`) que correspondiam à pergunta sobre receitas doces. Em seguida, ela injetou esses documentos como contexto no `rag_prompt`, permitindo que o LLM (`llm`) gerasse uma resposta informada, utilizando as informações extraídas dos PDFs.
